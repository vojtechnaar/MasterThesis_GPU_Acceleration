## Rule format note

`body_test_data/rule.txt` contains only the body atoms with short predicates — the C++ parser requires a full `body => head` line with prefixed predicate names that match the TTL.

The cell below creates `rule_formatted.txt` with:
- `rel:interacts_with` (parser reads prefixes from TTL and expands to full URI)
- Completed head atom `(?a rel:interacts_with ?b)`

```python
# Create properly formatted rule file
rule = "(?d rel:interacts_with ?c) ^ (?d rel:interacts_with ?b) ^ (?a rel:interacts_with ?e) => (?a rel:interacts_with ?b)\n"
with open("../body_test_data/rule_formatted.txt", "w") as f:
    f.write(rule)
print("Written rule_formatted.txt:")
print(rule)
```

## Run on first 100 000 triples from gpu-test.tsv

Takes the first 100 000 lines of `gpu-test.tsv`, converts to Turtle,
then evaluates the same rule on that smaller graph.

Lets you directly compare CPU vs GPU on identical data.

```python
import pathlib, csv, re

tsv_src = pathlib.Path('../body_test_data/gpu-test.tsv')
out_ttl = pathlib.Path('../body_test_data/gpu-test-100k.ttl')
N_TRIPLES = 100_000

ttd_drug_base = 'https://identifiers.org/ttd.drug/'
uniprot_base  = 'http://purl.uniprot.org/uniprot/'
rel_base      = 'https://example.org/relation/'
res_base      = 'https://example.org/resource/'

UNIPROT_RE = re.compile(
    r'^(?:[OPQ][0-9][A-Z0-9]{3}[0-9]|[A-NR-Z][0-9](?:[A-Z][A-Z0-9]{2}[0-9]){1,2})(?:-[0-9]+)?$')

def entity_ttl(tok):
    tok = tok.strip()
    if tok.startswith('ttd.drug_'):
        return f'ttd_drug:{tok[9:]}'
    if UNIPROT_RE.fullmatch(tok):
        return f'uniprot:{tok}'
    return f'res:{tok}'

lines_written = 0
with tsv_src.open(newline='') as fin, out_ttl.open('w') as fout:
    fout.write(f'@prefix ttd_drug: <{ttd_drug_base}> .\n')
    fout.write(f'@prefix uniprot:  <{uniprot_base}> .\n')
    fout.write(f'@prefix rel:      <{rel_base}> .\n')
    fout.write(f'@prefix res:      <{res_base}> .\n\n')
    reader = csv.reader(fin, delimiter='\t')
    for row in reader:
        if len(row) != 3 or not any(c.strip() for c in row):
            continue
        s, p, o = row
        fout.write(f'{entity_ttl(s)} rel:{p.strip()} {entity_ttl(o)} .\n')
        lines_written += 1
        if lines_written >= N_TRIPLES:
            break

print(f'Written {lines_written} triples to {out_ttl}')
print(f'File size: {out_ttl.stat().st_size / 1024:.1f} KB')
```

```python
!./bench_gpu ../body_test_data/gpu-test-100k.ttl ../body_test_data/rule_formatted.txt
```

```python
## Step 1: Recompile with -lineinfo
!nvcc -O3 -lineinfo -std=c++17 bench_gpu.cu FinalRule.cpp RdfIndexes.cpp RuleParser.cpp -o bench_gpu_profiled
```
